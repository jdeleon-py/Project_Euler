<p>The cube, $41063625$ ($345^3$), can be permuted to produce two other cubes: $56623104$ ($384^3$) and $66430125$ ($405^3$). In fact, $41063625$ is the smallest cube which has exactly three permutations of its digits which are also cube.</p>
<p>Find the smallest cube for which exactly five permutations of its digits are cube.</p>

```python
# define a function determining whether two given integers are permuted multiples of each other
def is_perm(x1: int, x2: int) -> bool:
    arr1 = np.sort(np.array([int(i) for i in str(x1)]))
    arr2 = np.sort(np.array([int(i) for i in str(x2)]))
    return np.array_equal(arr1, arr2)
```

```python
from collections import defaultdict

def func(target: int = 5) -> tuple[int, list[int]]:
    """
    - Return the smallest cube with exactly `target` cube permutations.
    - Checks groups only after finishing a full cube digit-length range.
    """
    if target < 1:
        raise ValueError("target must be positive")

    grps: dict[str, list[int]] = defaultdict(list)
    curr_dig: int = 1
    n: int = 1

    while True:
        cube: int = n ** 3
        cube_text: str = str(cube)
        digits: int = len(cube_text)

        if digits > curr_dig:
            candidates: list = [vals for vals in grps.values() if len(vals) == target]
            if candidates:
                smallest_grp: list = min(candidates, key=lambda vals: vals[0])
                return smallest_grp[0], smallest_grp
            grps.clear()
            curr_dig = digits

        grps["".join(sorted(cube_text))].append(cube)
        n += 1
```

```python
answer, permutations = func(5)
print("Smallest cube:", answer)
print("Cube permutations:", permutations)
print("Cube roots:", [round(c ** (1 / 3)) for c in permutations])
```

Runtime is approximately `O(N · d log d)` for `N` cubes of at most `d` digits; sorting at most 12 digits here is cheap.
